```python

import pandas as pd

data = {
    "Student": ["Aman", "Riya", "Rahul", "Neha", "Priya"],
    "Study_Hours": [2, 4, 3, 5, 4],
    "Math": [60, 85, 70, 90, 88],
    "Science": [65, 80, 75, 92, 86],
    "English": [70, 88, 72, 85, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python

print("Total students:", len(df))
print("Columns:", df.columns.tolist())
print("Average Math marks:", df["Math"].mean())
```

```python
print("Total students:", len(df))

print("Average Math marks:", df["Math"].mean())

print("Average Science marks:", df["Science"].mean())

print("Average English marks:", df["English"].mean())
```

```python
print("Highest Math marks:", df["Math"].max())

print("Lowest Math marks:", df["Math"].min())

print("Highest marks in Science:", df["Science"].max())

print("Lowest marks in Science:", df["Science"].min())
```

```python
top_student = df.loc[df["Math"].idxmax()]

print("Top student in Math:")
print("Name:", top_student["Student"])
print("Marks:", top_student["Math"])
```

```python
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English"]

averages = [
    df["Math"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

plt.bar(subjects, averages)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python
print("1. Total students:", len(df))

print("2. Best average subject:",
      df[["Math", "Science", "English"]].mean().idxmax())

print("3. Lowest average subject:",
      df[["Math", "Science", "English"]].mean().idxmin())

print("4. Top Math student:",
      df.loc[df["Math"].idxmax(), "Student"])
```

```python
df["Average_Marks"] = df[["Math", "Science", "English"]].mean(axis=1).round(2)

print(df[["Student", "Average_Marks"]])
```

```python
df.to_csv("student_performance.csv", index=False)

print("CSV file saved successfully!")
```
